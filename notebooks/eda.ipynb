# Exploratory Data Analysis - Hate Speech Detection Dataset

This notebook performs comprehensive EDA on the hate speech dataset including:
- Class distribution analysis
- Text length statistics
- Common words in hate vs non-hate speech
- Class imbalance detection
- Visualizations using matplotlib and seaborn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
from pathlib import Path
import sys

# Add project root to path
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

# Configure visualization style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 10

print("✓ Libraries imported successfully")
```

## 1. Load Dataset

```python
# Load dataset
data_path = project_root / 'data' / 'raw' / 'combined_hate_speech_dataset.csv'

if data_path.exists():
    df = pd.read_csv(data_path)
    print(f"✓ Dataset loaded: {data_path}")
else:
    # Create sample dataset for demonstration
    print("⚠ Dataset file not found. Creating sample dataset for demonstration...")
    np.random.seed(42)
    sample_texts_hate = [
        "You are so stupid", "I hate you all", "Go die", "Seriously hate this",
        "You should die", "This is disgusting", "I despise you", "You are pathetic",
        "Kill yourself", "You are worthless", "I absolutely hate you", "You are garbage"
    ]
    sample_texts_non_hate = [
        "I love this movie", "Great day today", "This is amazing", "Wonderful experience",
        "Beautiful sunset", "Amazing work", "Have a nice day", "Fantastic performance",
        "Excellent job", "This is wonderful", "I really enjoyed it", "Beautiful day"
    ]
    
    df = pd.DataFrame({
        'text': sample_texts_hate + sample_texts_non_hate,
        'label': [1] * len(sample_texts_hate) + [0] * len(sample_texts_non_hate)
    })
    print(f"✓ Sample dataset created with {len(df)} texts")

print(f"\nDataset Shape: {df.shape}")
print(f"\nFirst few rows:")
print(df.head())
print(f"\nData types:")
print(df.dtypes)
print(f"\nMissing values:")
print(df.isnull().sum())
```
